```python
only_female = insurance[insurance.sex == "female"]
print(only_female)
```

```python
only_female_df = pd.DataFrame(only_female)
print(only_female_df)
```

```python
cost_vs_child_histplot = sns.boxplot(x="children", y="charges", data=only_female_df)
plt.title("Insurance Rate and Number of Children")
plt.savefig("cost and children.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

On average, females who have 1 child pay less insurance cost than those who have none, however those with 2-4 children pay more, with the average dipping below that of having no kids again when they have 5. There are plenty of outliers however, which are likely due to other factors such as smoking, which are skewing the numbers. This graph also cannot tell us how many females per number of children there are, which could also be skewing results due to categories not being the same size.

```python
bmi_vs_children = sns.boxplot(x="children", y="bmi", data=only_female_df)
plt.title("BMI and Number of Children")
plt.savefig('bmi and children.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```

BMI does not have a widely differing average for females, regardless of how many children they may have, although those with two children have aa higher average BMI than those with less or no children or those with more children.

```python
import csv
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
insurance = pd.read_csv("insurance.csv")
```

```python
print(insurance.head())
```

```python
cost_by_sex = insurance.groupby("sex").charges.mean()
print(cost_by_sex)
```

```python
13956.75-12569.57
```

```python
#regardless of other factors, there is a $1,387.18 difference in average insurance costs between females and males, with males spending more on average
```

```python
only_female = insurance[insurance.sex == "female"]
print(only_female)
```

```python
only_female_df = pd.DataFrame(only_female)
print(only_female_df)
```

```python
cost_vs_child_histplot = sns.boxplot(x="children", y="charges", data=only_female_df)
plt.show()
plt.close()
print(cost_vs_child_boxplot)
```

```python
#on average, females who have 1 child pay less insurance cost than those who have none, however from 2-4 children pay more, with the average dipping below that of having no kids again when there are 5. there are plenty of outliers however, which are likely due to other factors such as smoking, which are skewing the numbers. this graph also cannot tell us how many females per number of children there are, which could also be skewing results due to categories not being the same size.
```
